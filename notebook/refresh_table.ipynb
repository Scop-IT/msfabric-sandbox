```python
import sempy.fabric as fabric

# ─────────────────────────────────────────────────────────────
#  CONFIG — edit these (everything is hardcoded, no prompts)
# ─────────────────────────────────────────────────────────────
WORKSPACE    = "My Workspace"  # workspace name
DATASET      = "My Semantic Model Name"     # semantic model name
TABLE        = "My Table"                 # table to refresh
REFRESH_TYPE = "full"                   # full | dataOnly | calculate | clearValues | automatic | defragment

# ─────────────────────────────────────────────────────────────
#  REFRESH
# ─────────────────────────────────────────────────────────────
print(f"Refreshing [{TABLE}] in '{DATASET}' ({WORKSPACE}) — type={REFRESH_TYPE}")

fabric.refresh_dataset(
    dataset=DATASET,
    workspace=WORKSPACE,
    objects=[{"table": TABLE}],
    refresh_type=REFRESH_TYPE,
)

print("Refresh triggered successfully.")
```
