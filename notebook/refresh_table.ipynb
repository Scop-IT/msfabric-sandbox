```python
# (Optional) Install dependencies if missing from the runtime.
# Fabric runtimes ship PyJWT 2.4.0, but fsspec-wrapper (pulled by sempy_labs)
# requires PyJWT >= 2.6.0 — so we bump it here to avoid the dependency conflict.
# You can delete this cell if your runtime already has both at the right versions.
import importlib.metadata as md
import importlib.util
import notebookutils

need_restart = False

# PyJWT >= 2.6.0
try:
    pyjwt_ok = tuple(int(x) for x in md.version("PyJWT").split(".")[:2]) >= (2, 6)
except Exception:
    pyjwt_ok = False
if not pyjwt_ok:
    %pip install -q "PyJWT>=2.6.0"
    need_restart = True

# semantic-link-labs
if importlib.util.find_spec("sempy_labs") is None:
    %pip install -q semantic-link-labs
    need_restart = True

if need_restart:
    print("Dependencies installed/updated, restarting Python...")
    notebookutils.session.restartPython()
else:
    print("Dependencies already satisfied.")
```
